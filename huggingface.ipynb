IMPORTING PIPELINE


```python
from transformers import pipeline


#------------- NLP TASK -----------------#

'''
1. Text Classification:
Assigning a category to a given text.
Sentiment Analysis: Classifying text as positive, negative, or neutral.
Topic Classification: Assigning a text to a specific topic or category.
Spam Detection: Identifying whether an email is spam or not.
Language Detection: Detecting the language of a given text.
'''
classifier = pipeline("text-classification")


'''
2. Token Classification:
Assigning label to individula tokens in a sequence
Named Entity Recognition (NER)
Part-of-speach Tagging

'''
token_classifier = pipeline('token-classification')



'''
3. Question Answering: Answering question

'''
question_answerer = pipeline('question-answering')



'''
4. Summarization: Generating a summary of a given text

'''
summarizer = pipeline('summarization')



'''
5. Translation: Translating text from one language to another

'''
translator = pipeline('translation', model='HelpMumHQ/AI-translator-eng-to-9ja')



'''

6. Text Generation: Generating new text based on a given prompt

'''
generator = pipeline('text-generation')



'''
7. text2text Generation: Generating and summarization of new text based on a given prompt

'''
text2text_generator = pipeline('text2text-generation')



'''
8. Fill-Mask: Filling a mask in a given text

'''
fill_mask = pipeline('fill-mask')



'''
9. Feature Extraction: Extracting features from a given text

'''
feature_extractor = pipeline('feature-extraction')



''''
10. Sentence Similarity: Measuring the similarity between two sentences

'''
similarity = pipeline('sentence-similarity')



'''
11. Image Classification: Assigning a label to an image

'''
image_classifier = pipeline('image-classification')



'''
12. Image Segmentation: Assigning a label to each pixel in an image

'''
image_segmenter = pipeline('image-segmentation')




'''
13. Image Captioning: Generating a caption for an image

'''
image_captioner = pipeline('image-captioning')


'''
14. Image Generation: Generating a new image based on a given prompt

'''
image_generator = pipeline('image-generation')



'''
15. Object Detection: Identifying and localizing objects in an image

'''
object_detector = pipeline('object-detection')





#==================== Speach Processing Tasks ====================#

'''
1. Automatic Speech Recognition (ASR): Converting speech to text.
Automatic

'''
speech_recognizer = pipeline('automatic-speech-recognition')


'''

2. Audio Clissification
3. Speach to Speech Translation
4. Text to Speech Conversion

'''

#******************** Multimodal Tasks ********************#

'''
1. Image Captioning: Generating a caption for an image.

'''
image_captioner = pipeline('image-captioning')


'''
2. Visual Question Answering (VQA): Answering questions about images.

'''
vqa = pipeline('visual-question-answering')


#>>>>>>>>>>>>>>>>>>>>>>>>>>>> Other Tasks >>>>>>>>>>>>>>>>>>>>>>>>>>>>#

'''
Table Question Answering: Answering questions about tables of data.

'''
table_qa = pipeline('table-question-answering')





```

EXAMPLES

1.Sentiment Analysis


```python
from transformers import pipeline

classifier = pipeline("sentiment-analysis", model='distilbert/distilbert-base-uncased-finetuned-sst-2-english')
classifier("I've been waiting for a HuggingFace course my whole life.")
```

```python
# One line code
pipeline("sentiment-analysis")("I've been waiting for the country to change for the better.")
```

```python
#Passing multiple, usa array

pipeline('sentiment-analysis', model='distilbert/distilbert-base-uncased-finetuned-sst-2-english')(['I love you', 'I hate you', 'I dont know', 'if you say so', 'I do not mind'])
```

```python
#Passing multiple, usa array

pipeline('sentiment-analysis', model='SamLowe/roberta-base-go_emotions')(['I love you', 'I hate you', 'I dont know', 'if you say so', 'I do not mind'])
```

QUESTION AND ANSWER

```python
from transformers import pipeline

question_answerer = pipeline("question-answering", model='distilbert/distilbert-base-cased-distilled-squad')
my_ans = question_answerer(
    question="Where do I work?",
    context="My name is Sylvain and I work at Tele.Com in Brooklyn",
)

print(my_ans)
```
